# Reading CSVs in Detail

`read_csv` is the workhorse loader. This notebook focuses on the parameters that reshape data *while loading it*, then on Series-level selection and summary methods.

```python
import pandas as pd
```

```python
# Local execution already has the data files.
# On Colab (which opens just this notebook) we download them:
import os
import urllib.request

NEEDS_FILES = ['pokemon.csv', 'google_stock_price.csv']

for _f in NEEDS_FILES:
    if not os.path.exists(_f):
        urllib.request.urlretrieve('https://raw.githubusercontent.com/sivabalanb/Data-Analysis-with-Pandas-and-Python/master/' + _f, _f)
        print('downloaded', _f)
```

## `usecols` — load only what you need

`usecols=["Pokemon"]` loads a single column. `["col"]["col"]` then extracts it as a Series (the old `squeeze=True` parameter was removed — this is its modern replacement).

```python
pokemon = pd.read_csv("pokemon.csv", usecols=["Pokemon"])["Pokemon"]
pokemon
```

```python
google = pd.read_csv("google_stock_price.csv")["Stock Price"]
google
```

## Sorting and naming

- `sort_values()` sorts the values; `ascending=False` reverses.
- `inplace=True` modifies the object instead of returning a copy.
- `.name = "..."` labels the Series.

```python
pokemon.sort_values().head()
```

```python
pokemon.name = "Pokemon Header"
```

```python
pokemon.sort_values(ascending= False)
```

```python
google.sort_values()
```

```python
# Selecting a column returns a view; copy before sorting in place (pre-2.0 pandas requirement).
google = google.copy()
google.sort_values(ascending=False, inplace=True)
```

```python
google.head()
```

## Indexing and lookup

- `pokemon[4]` — positional access by label if index is 0..n-1 (here it is).
- `read_csv(index_col="Pokemon")` sets a column as the index during load.
- `.get("Hoopa")` — safe lookup that returns `None` (not an error) for missing labels.

```python
pokemon = pd.read_csv("pokemon.csv", index_col="Pokemon")["Type"]
pokemon
```

```python
pokemon.get("Hoopa")
pokemon.get(key = "Noibat")
```

## Two columns: `map`

With a name column and a type column, `pokemon_names.map(pokemon_types)` looks up each name's type — the building block of lookups and joins.

```python
pokemon_names = pd.read_csv("pokemon.csv", usecols=["Pokemon"])["Pokemon"]

pokemon_types = pd.read_csv("pokemon.csv", index_col="Pokemon")["Type"]
pokemon_names
pokemon_types
```

```python
pokemon_names
```

```python
pokemon_types
```

```python
pokemon_names.map(pokemon_types)
```

```python
pokemon_names = pd.read_csv("pokemon.csv", usecols=["Pokemon"])["Pokemon"]

pokemon_types = pd.read_csv("pokemon.csv", index_col="Pokemon")["Type"].to_dict()
pokemon_types
```
