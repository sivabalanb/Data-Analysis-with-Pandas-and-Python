## A wide, tall DataFrame

`np.random.randint(...)` builds a 1000x50 DataFrame — too big to print fully. Pandas truncates the display by default. That's what the options below control.

```python
data = np.random.randint(1,100,[1000,50])
```

```python
df = pd.DataFrame(data)

df.tail()
```

## Display options

- `pd.options.display.max_rows` — cap on printed rows.
- `pd.options.display.max_columns` — cap on printed columns.

Setting `max_rows = 2` shows only the first and last row with an ellipsis.

```python
pd.options.display.max_rows
```

```python
pd.options.display.max_rows = 2
```

```python
df
```

```python
pd.options.display.max_columns = 5
```

```python
df
```

## `get_option` / `set_option`

The function API is equivalent to the `options` object: `pd.get_option("display.max_columns")` reads, `pd.set_option(...)` writes. `pd.describe_option(...)` explains what an option does. `pd.reset_option(...)` restores defaults.

```python
pd.get_option("display.max_columns")
```

```python
pd.set_option("display.max_columns", 20)
```

```python
pd.reset_option
```

```python
pd.describe_option("display.max_columns")
```

## `precision`

`pd.set_option("display.precision", 2)` rounds *displayed* floats to 2 decimals. The stored values are unchanged — only the printed representation is rounded.

```python
pd.DataFrame(np.random.randn(5,5))
```

```python
pd.get_option("display.precision")
```

```python
pd.set_option("display.precision", 2)
```
