# The Series

A **Series** is a one-dimensional labeled array — a column of values, each with a name (its *index*).

Think of it as a Python list with a sticky label on every element. Everything you'll learn here is the foundation for DataFrames.

Create series in python

## Creating a Series

Pass a list to `pd.Series()` and pandas builds a Series with a default integer index (`0, 1, 2, ...`). Note how the data type is inferred from the values.

```python
import pandas as pd
ice_cream = ['vanilla','chocolate','mousse','rocky road']
```

```python
merriam = {"Work":"Office","Play":"Stadium","Study":"Class"}
pd.Series(merriam)
```

## Attributes

Attributes are properties of the object (no parentheses). They let you inspect the raw data, the index, and the data type.

```python
## Intro to Attributes
```

```python
about_me = ["Yo","Yep","thats","awesomwe"]
s = pd.Series(about_me)
s
```

```python
prices = [1.44,6.8,9.65]
s = pd.Series(prices)
s
```

## Methods

Methods are actions the Series can perform (called with parentheses). `sum()` and `mean()` work because the values are numeric. Try calling `.sum()` on a Series of strings to see why types matter.

```python
s.sum()
```

```python
s.mean()
```
