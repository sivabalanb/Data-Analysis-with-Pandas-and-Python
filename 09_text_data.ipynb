# Working with Text

The `.str` accessor applies **string methods to every value** in a column at once — no `for` loop. It mirrors Python's string methods: `.lower()`, `.title()`, `.len()`, `.strip()`, `.replace()`, `.split()`, ...

## Vectorized string methods

- `"HELLO WORLD".lower()` — the plain Python baseline.
- `chicago["Name"].str.lower()` — the same thing, applied to a whole column.

Every normal string method has a `.str.` counterpart.

```python
"HELLO WORLD".lower()
```

```python
chicago["Name"].str.lower()
```

```python
chicago["Name"].str.title()
```

## Cleaning real data

The salary column is text (`$90744.00`). `str.replace("$", "")` strips the dollar sign, then `astype(float)` makes it numeric. This is the single most common text-cleaning task in real jobs.

```python
chicago["Department"].str.len()
```

```python
"Hello World".replace("l","t")
```

```python
chicago.head()
```

```python
chicago["Employee Annual Salary"] = chicago["Employee Annual Salary"].str.replace("$","").astype(float)
```

```python
chicago
```

## Cleaning the index

`str.strip()` trims whitespace, `.str.title()` fixes casing — chained together and reassigned to the index. `str.split(",").str.get(0)` splits on the comma and keeps the first piece (the last name).

```python
chicago["Name"].str.strip()
```

```python
chicago = pd.read_csv("chicago.csv", index_col="Name").dropna(how = "all")
chicago["Department"] = chicago["Department"].astype("category")
chicago.head()
```

```python
chicago.index.nunique()
```

```python
chicago.index = chicago.index.str.strip().str.title()
```

```python
chicago.columns = chicago.columns.str.title()
```

```python
chicago.head()
```

## When to use `.str`

Use `.str` for per-character transformations. For counting, matching, or replacing patterns, learn `str.contains`, `str.extract`, and `str.replace` with regex — they follow in an advanced notebook.

```python
chicago = pd.read_csv("chicago.csv").dropna(how = "all")
chicago["Department"] = chicago["Department"].astype("category")
chicago.head()
```

```python
chicago["Name"].str.split(",").str.get(0).str.title().value_counts()
```
