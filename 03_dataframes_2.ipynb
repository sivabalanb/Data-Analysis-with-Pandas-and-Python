## Setting up types

- `parse_dates=[...]` tells pandas which columns are dates during load.
- Columns are converted to their natural types: `bool` and `category`.

Getting types right up front prevents a whole class of bugs later.

```python
df = pd.read_csv("employees.csv", parse_dates=["Start Date", "Last Login Time"] )
df["Senior Management"] = df["Senior Management"].astype("bool")
df["Gender"] = df["Gender"].astype("category")
```

## Duplicates

- `.duplicated()` marks rows that repeat.
- `~` inverts a boolean mask (`~duplicated` = *not* duplicated).
- `drop_duplicates(subset=..., keep=False)` removes duplicate rows, keeping none of the copies.

The `keep=False` option is useful for finding names that are *shared* between employees — a common data-quality check.

```python
mask = ~df["First Name"].duplicated(keep = False)
```

```python
df[mask]
```

```python
len(df.drop_duplicates(subset = "First Name", keep = False))
```
