# GroupBy: Split, Apply, Combine

The single most important pattern in data analysis:

1. **Split** the data into groups (e.g., by Sector).
2. **Apply** a function to each group independently (sum, mean, max...).
3. **Combine** the results into one table.

This notebook works on the Fortune 1000 list — every company, its sector, revenue and profits.

## Creating a GroupBy object

`fortune.groupby("Sector")` splits the data into one group per sector. It is **lazy** — nothing computes until you apply an operation.

- `len(sectors)` — number of groups
- `.nunique()` — number of distinct sectors (should match)

```python
fortune = pd.read_csv("fortune1000.csv", index_col="Rank")
sectors = fortune.groupby("Sector")
fortune.head()
```

## Getting one group

`get_group("Energy")` returns the full DataFrame for one group. It's the fastest way to inspect a single slice of the data.

```python
sectors.get_group("Energy")
sectors.get_group("Technology")
```

## Aggregating every group

`max()`, `min()`, `sum()` run on every group. Note the output — one row per sector.

`sum()` is used on text too; only later we'll learn `agg` for fine-grained control.

```python
sectors.max()
sectors.min()
sectors.sum()
```

```python
sectors.get_group("Apparel")["Profits"].sum()
```

## Grouping by multiple columns

`groupby(["Sector", "Industry"])` creates one group for each *combination*. The result is a MultiIndex (see the MultiIndex notebook).

```python
fortune = pd.read_csv("fortune1000.csv", index_col="Rank")
sectors = fortune.groupby(["Sector","Industry"])
fortune.head()
```

```python
sectors.sum()
```

```python
fortune = pd.read_csv("fortune1000.csv", index_col="Rank")
sectors = fortune.groupby("Sector")
fortune.head()
```

## `agg` — different operations per column

`.agg({"Revenue": "sum", "Profits": "sum", "Employees": "mean"})` applies a different function to each column. This is the workhorse for real reporting.

```python
sectors.agg({"Revenue":"sum",
            "Profits":"sum",
            "Employees":"mean"})
```

## Iterating groups

`for sector, data in sectors:` loops over `(group_name, group_data)` pairs. Here we keep the single highest-revenue company from each sector and collect them with `pd.concat`.

```python
df = pd.DataFrame(columns= fortune.columns)
```

```python
df = pd.DataFrame()
for sector, data in sectors:
    highest_revenue = data.nlargest(1, "Revenue")
    df = pd.concat([df, highest_revenue])
df
```

```python
df
```

## Grouping by location

Same pattern, new grouping key (`Location`). Notice the pattern generalizes: change the key, keep the logic.

```python

cities = fortune.groupby("Location")
```

```python
df = pd.DataFrame()
for sector, data in cities:
    highest_revenue = data.nlargest(1, "Revenue")
    df = pd.concat([df, highest_revenue])
df
```
