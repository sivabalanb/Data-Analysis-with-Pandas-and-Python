## Before you start: two concepts that unlock everything

### 1. Two kinds of containers
- **Series** — a single labeled column of values (like a list with names attached).
- **DataFrame** — a table of rows and columns (like a spreadsheet). A DataFrame is made of Series.

### 2. The index
Every Series and DataFrame has an **index** — the row labels. The index can be integers (`0, 1, 2, ...`), dates, names, anything. Two of the most common beginner bugs come from confusing **labels** (`loc`) with **positions** (`iloc`). You will see both everywhere in the next notebooks.

## How to use this repository
- **Follow along**: run every cell, then change the values to see what breaks. Breaking things is how you learn Pandas.
- **Look at the data first**: before writing any analysis, always run `df.head()`, `df.info()`, and `df.describe()`.
- **Ask 'what problem does this solve?'**: every method here exists because real analysis needs it. The notebooks call out the *why*.
