# Input and Output

Reading and writing data — the boring 20% that unblocks the interesting 80%. Focus here on the options that *change the file you get out*.

```python
import pandas as pd
```

```python
# Local execution already has the data files.
# On Colab (which opens just this notebook) we download them:
import os
import urllib.request

NEEDS_FILES = ['crime_india.csv', 'Data - Multiple Worksheets.xlsx']

for _f in NEEDS_FILES:
    if not os.path.exists(_f):
        urllib.request.urlretrieve('https://raw.githubusercontent.com/sivabalanb/Data-Analysis-with-Pandas-and-Python/master/' + _f, _f)
        print('downloaded', _f)
```

## `read_csv`

`read_csv` can read a local file or a URL. `.head()` shows a preview. The file `crime_india.csv` was fetched from India's open-data portal and saved into this repo so the notebook runs offline.

## `to_csv` — writing

- By default the index is written (as an extra column). `index=False` drops it.
- For text-heavy data, `encoding="UTF8"` avoids encoding errors.

```python
t.to_csv("test_Export.csv")
```

```python
t.to_csv("test_Export.csv",index=False)
```

## `read_excel` — multi-sheet files

- No argument: first sheet.
- `sheet_name=1`: second sheet (0-based).
- `sheet_name=[0, 1]`: a list returns a **dict of DataFrames**, keyed by sheet.

```python
pd.read_excel("Data - Multiple Worksheets.xlsx")
```

```python
pd.read_excel("Data - Multiple Worksheets.xlsx", sheet_name=1)
```

```python
pd.read_excel("Data - Multiple Worksheets.xlsx", sheet_name=[0,1])
```

## Writing Excel

`pd.ExcelWriter` opens an Excel file; `to_excel` writes a DataFrame into it. The `with` block ensures the file is saved and closed even if code fails — always use it.

```python
state = crime["State/UT (Col.3)"]
state.head()
```

```python
with pd.ExcelWriter("State Name.xlsx") as excel_file:
    state.to_excel(excel_file, index=False)
```
