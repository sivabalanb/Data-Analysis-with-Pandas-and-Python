## `pd.Timestamp` — pandas' date type

`pd.Timestamp` parses *many* string formats automatically. Notice that `"1/6/2015"` is read as January 6th (US order) — always verify the format, this is a classic source of wrong results.

```python
pd.Timestamp('2015-1-6')
pd.Timestamp('2015/1/6')
pd.Timestamp('1/6/2015')
pd.Timestamp('12/19/2015')
pd.Timestamp('19/12/2015')
```

## `DatetimeIndex` — dates as labels

A list of date strings becomes a `DatetimeIndex`, usable as a Series index. Slicing by date then *just works*.

```python
dates = ['2015-1-6','2015-5-6','2015-2-2']
dtIndex = pd.DatetimeIndex(dates)
```

```python
values= [100,200,400]
```

```python
pd.Series(data = values , index=dtIndex)
```

## `pd.date_range` — generate dates

`freq=` controls the spacing:

- `D` daily, `W` weekly (or `W-FRI` to fix the day), `5h` every 5 hours
- `ME` month-end, `YE` year-end (modern aliases; the old `M`/`A` are gone)

Use `periods=` instead of `end=` to request a specific count.

```python
times = pd.date_range(start='2016', end='2018',freq='D')
```

```python
type(times[0])
```

```python
times = pd.date_range(start='2016', end='2018', freq='W')
```

```python
times
```

```python
pd.date_range(start='2016', end='2018', freq='W-FRI')
```

```python
pd.date_range(start='2016', end='2018',freq='5h')
```

```python
pd.date_range(start='2016', end='2018', freq='ME')
```

```python
pd.date_range(start='2016', end='2018', freq='YE')
```

```python
pd.date_range(start='2014', periods=25)
```

```python
bunch_of_dates = pd.date_range(start='2000',end='2016', freq='2D')
```

## Dates as an index

Here a daily price series (generated locally, so no network needed) uses a `DatetimeIndex`. That enables label selection with strings: `stocks.loc["2014-04-01"]`.

`isin(birthday)` finds all rows whose date matches any date in another range — e.g. extracting birthdays from daily data.

```python
import numpy as np
np.random.seed(42)
rng = pd.date_range("2010-01-01", "2016-12-12", freq="D")
price = 50 + np.cumsum(np.random.normal(0, 1, len(rng)))
stocks = pd.DataFrame({
    "Open": price, "High": price + 1, "Low": price - 1,
    "Close": price + np.random.normal(0, 0.5, len(rng)),
    "Volume": np.random.randint(1_000_000, 5_000_000, len(rng)),
}, index=rng)
```

```python
stocks.head()
```

```python
stocks.loc["2014-04-01"]
stocks.iloc[299]
```

```python
stocks.loc['2014-04-01']
stocks.iloc[299]
```

```python
birthday = pd.date_range(start='2000-04-01', end='2016-04-01', freq=pd.DateOffset(years=1))
```

```python
mask = stocks.index.isin(birthday)
```

```python
stocks.loc[mask]
```

## Date arithmetic

- `truncate(before, after)` slices by date labels.
- `index + pd.DateOffset(months=4)` shifts every date forward (calendar-aware).
- `MonthEnd()` rounds to month boundaries.
- Subtracting two `Timestamp`s gives a `Timedelta` — the elapsed time.

```python
stocks.truncate(before="2009-12-31", after="2011-12-31")
```

```python
stocks.index + pd.DateOffset(months = 4)
```

```python
stocks.index+ pd.tseries.offsets.MonthEnd()
```

```python
timeA = pd.Timestamp('2010-01-31')
timeB = pd.Timestamp('2016-01-31')
```

```python
timeA - timeB
```

```python
type(timeA - timeB)
```

## `Timedelta` — durations

`pd.Timedelta(days=4, weeks=8, hours=12)` builds a duration; you can also parse a string like `"6 hours 12 minutes"`.

```python
pd.Timedelta(days = 4, weeks = 8, hours = 12)
```

```python
pd.Timedelta("6 hours 12 minutes")
```

## Real example: delivery times

The `ecommerce.csv` data has an order date and a delivery date. Subtracting them computes `delivery_time`. Comparing against a duration (`> '256 days'`) finds the slowest orders — the pattern of a real logistics report.

```python
shipping = pd.read_csv('ecommerce.csv', index_col='ID', parse_dates=['order_date','delivery_date'])
```

```python
shipping['delivery_time'] = shipping['delivery_date'] - shipping['order_date']
```

```python
shipping.head()
```

```python
mask = shipping['delivery_time'] > '256 days'
shipping[mask]
```
