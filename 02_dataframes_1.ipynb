## Rows and columns: the `axis` trap

`sum()` without arguments sums down each *column*. `sum(axis=1)` sums across each *row*. The #1 beginner confusion in pandas is `axis`: `axis=0` = columns (default), `axis=1` = rows.

```python
nba.dtypes.value_counts()
```

```python
rev = pd.read_csv("revenue.csv", index_col= "Date")
```

```python
rev.sum()
```

```python
rev.sum(axis = 1 )
```

## Missing data

Real data has holes. Two ways to handle them:

- `dropna()` — drop rows with missing values (default) or all-empty rows with `how="all"`.
- `fillna(value)` — fill holes with a default (0 for numbers, `"None"` for text).

There is no universally correct choice — it depends on what the data means.

```python
nba.dropna().tail()
```

```python
nba.dropna(how = "all")
```

```python
import pandas as pd
nba = pd.read_csv("nba.csv").dropna(how = "all")

nba["Salary"].fillna(0, inplace = True)
nba["College"].fillna("None", inplace = True)
```

```python
nba.tail()
```

## Column types: `dtype` matters

`astype()` converts a column's type. This is required before operations that assume numbers. Note that converting to `int` fails if the column still contains `NaN` — that's why we `dropna` first above.

```python
nba.dtypes

nba.info()
```

```python
nba = nba.dropna(subset=["Salary", "Number", "Age"])
nba["Salary"] = nba["Salary"].astype("int")
```

```python
nba["Number"] = nba["Number"].astype("int")
nba["Age"] = nba["Age"].astype("int")
```

## Categories

`astype("category")` stores repeated strings (like Team names) more compactly. It also records the set of allowed values, which is useful for consistency checks.

```python
nba["Position"].nunique()
```

```python
nba["Position"] = nba["Position"].astype("category")
```

```python
nba["Team"].nunique()
```

```python
nba["Team"] = nba["Team"].astype("category")

```

## Sorting

- `sort_values("Salary")` sorts by a column.
- `sort_index()` sorts by the row labels.

`rank()` assigns each value its position in the ordering — here, the salary rank from highest to lowest.

```python
nba.sort_values("Name", ascending= False)
nba.sort_values("Salary", ascending= False)

nba.sort_index(ascending= False)
```

```python
nba["Salary Rank"] = nba["Salary"].rank(ascending = False).astype("int")
```

```python
nba.sort_values(by = "Salary", ascending = False)
```
